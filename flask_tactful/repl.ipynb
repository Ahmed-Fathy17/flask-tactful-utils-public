```python
from flask_tactful.ddd import Event
```

```python
from redis import Redis
redis = Redis.from_url(url="redis://localhost:6379")
print(redis.client_list())
```

```python
redis.xinfo_stream("tactful/billing")
```

```python
redis.xinfo_groups("tactful/billing")
```

```python
redis.xgroup_create(name="tactful/billing", groupname="client1",mkstream=True)
```

```python
redis.xgroup_createconsumer(name="tactful/billing", groupname="client1", consumername="pod2")
```

```python
ev = Event(topic="tactful/billing", profile_id=1)
redis.xadd("tactful/billing", ev.dict())
```

```python
ev_in = redis.xread({"tactful/billing":"0"}, count=1)
```

```python
redis.xreadgroup(groupname="client1", consumername="pod2", streams={"tactful/billing":">"}, count=1)
```

```python
redis.xreadgroup(groupname="client1", consumername="pod1", streams={"tactful/billing":">"}, count=1)
```
